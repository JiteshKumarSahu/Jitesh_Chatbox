```python
import json
import random
import joblib

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

# load dataset
with open("cht_boxdata.json","r") as file:
    data = json.load(file)


sentence=[]
label=[]
response={}

for i in data["intents"]:
    tag=i['tag']
    response[tag]=i["responses"]

    for pettern in i["patterns"]:
        sentence.append(pettern)
        label.append(tag)

print(sentence)
print(label)

#train test split

x_train,x_test,y_train,y_test = train_test_split(sentence,label)


#TF-TDF

vector= TfidfVectorizer()
x_train_n=vector.fit_transform(x_train).toarray()
x_test_n=vector.transform(x_test).toarray()

#train model
model = GaussianNB()
model.fit(x_train_n,y_train)

#Test Model
y_predict = model.predict(x_test_n)
score = accuracy_score(y_test, y_predict)
print("Model Accuracy: ", score)

# Save Model

joblib.dump(model,"Chatbot_Model.pkl")
joblib.dump(vector,"tfidf.pkl")
joblib.dump(response,"responses.pkl")
```
